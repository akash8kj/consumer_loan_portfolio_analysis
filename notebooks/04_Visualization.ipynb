```python
import pandas as pd
```

```python
df =pd.read_csv("Loans.csv")
```

```python
df_clean =df.copy()
```

```python
import matplotlib.pyplot as plt

df_clean['loan_status'].value_counts().plot(kind='bar')

plt.title('Loan Status Distribution')
plt.xlabel('Loan Status')
plt.ylabel('Number of Loans')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_clean['loan_purpose'].value_counts().plot(kind='bar')

plt.title('Loan Purpose Distribution')
plt.xlabel('Loan Purpose')
plt.ylabel('Number of Loans')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_clean.groupby('grade')['interest_rate'].mean().plot(kind='bar')
plt.title('Average Interest Rate by Loan Grade')
plt.xlabel('Loan Grade')
plt.ylabel('Average Interest Rate (%)')
plt.tight_layout()
plt.show()
```

```python
df_clean.groupby('loan_purpose')['loan_amount'].mean().sort_values().plot(kind='barh')
plt.title('Average Loan Amount by Purpose')
plt.xlabel('Average Loan Amount')
plt.ylabel('Loan Purpose')
plt.tight_layout()
plt.show()
```
